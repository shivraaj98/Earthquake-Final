```python
"""
Earthquake Magnitude Forecasting — Production-Grade Pipeline v2
===============================================================
FIX SUMMARY (vs original):
  [1] ENTROPY POOL:  All env vars set BEFORE any TF/CUDA import; GPU +
                     cuDNN determinism flags added; safe try/except guard.
  [2] LEAKAGE-FREE:  Temporal split is performed FIRST. Features are still
                     computed on the full ordered df (all operators are
                     strictly causal / left-sided), but the scaler is fit
                     exclusively on training rows.
  [3] LOW-PASS:      EMA retained + causal 3rd-order Butterworth IIR added
                     (steeper roll-off, no phase distortion on forward pass).
  [4] HIGH-PASS:     Butterworth residual replaces EMA residual (cleaner
                     frequency separation). EMA residual kept as extra feature.
  [5] INTER-EVENT:   log1p(hours) retained + cyclic sin/cos hour-of-day &
                     day-of-week encodings (captures diurnal/weekly patterns).
  [6] BENIOFF STRAIN: Cumulative √(10^1.5M) in a rolling window captures
                      seismic energy accumulation between major ruptures.
  [7] SEQUENCES:     Cross-split warm-up — val/test use the tail of the prior
                     split as context, so ALL val/test samples get predictions
                     (original lost SEQ_LEN samples per boundary).
  [8] SAMPLE WEIGHTS: log-linear weighting replaces exp() which was
                      numerically explosive (exp(5)≈148× for a M9 vs M4).
  [9] ARCHITECTURE:  causal Conv1D padding; Multi-Head Self-Attention with
                     residual + LayerNorm between BiLSTM layers.
  [10] SHUFFLE:      shuffle=False in model.fit() — temporal mini-batch order
                     is preserved (avoids spurious gradient patterns).
  [11] EVALUATION:   Per-bin stratified metrics, directional accuracy,
                     residual-vs-magnitude plot, 4-panel diagnostic figure.
"""

# ============================================================
# SECTION 0 — ENTROPY POOL
# Must be set BEFORE any TensorFlow / CUDA import.
# TF_DETERMINISTIC_OPS controls GPU kernel selection.
# TF_CUDNN_DETERMINISTIC forces deterministic cuDNN algorithms.
# ============================================================
import os
import random
import sys

SEED = 42
os.environ["PYTHONHASHSEED"]         = str(SEED)
os.environ["TF_DETERMINISTIC_OPS"]   = "1"   # GPU op-level determinism
os.environ["TF_CUDNN_DETERMINISTIC"] = "1"   # cuDNN algorithm pinning

# Uncomment the line below to force CPU-only execution for perfect
# reproducibility (GPUs have non-deterministic floating-point ordering
# even with the env vars above on some hardware):
# os.environ["CUDA_VISIBLE_DEVICES"] = "-1"

import numpy as np
random.seed(SEED)
np.random.seed(SEED)

import pandas as pd
import matplotlib
matplotlib.use("Agg")          # headless backend — safe in all environments
import matplotlib.pyplot as plt

from scipy.signal import butter, sosfilt   # causal Butterworth IIR

from sklearn.preprocessing import RobustScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import tensorflow as tf
tf.random.set_seed(SEED)

# enable_op_determinism() was added in TF 2.9; guard for older versions.
try:
    tf.config.experimental.enable_op_determinism()
    print("[OK] Op-level determinism enabled.")
except (AttributeError, RuntimeError) as exc:
    print(f"[WARN] enable_op_determinism unavailable ({exc}). "
          "Falling back to env-var determinism — reproducibility on GPU "
          "is best-effort only.")

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D, BatchNormalization, Dropout,
    Bidirectional, LSTM, Dense, MultiHeadAttention,
    GlobalAveragePooling1D, LayerNormalization, Add,
)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.regularizers import l2
from tensorflow.keras.losses import Huber


# ============================================================
# SECTION 1 — HYPER-PARAMETERS
# ============================================================
FILE_PATH    = "https://raw.githubusercontent.com/shivraaj98/Earthquake/main/earthquake.csv"
SEQ_LEN      = 21        # look-back window (steps)
TRAIN_RATIO  = 0.80
VAL_RATIO    = 0.10
# TEST_RATIO = 0.10 (remainder)

EPOCHS       = 200
BATCH_SIZE   = 32
LR           = 3e-4
L2_REG       = 0.003
DROPOUT      = 0.35
HUBER_DELTA  = 1.0       # Huber loss transition point (robust to outliers)


# ============================================================
# SECTION 2 — DATA LOADING & VALIDATION
# ============================================================
print("\n[1/10] Loading data …")
df = pd.read_csv(FILE_PATH)

# --- Robust datetime parsing ---
# Some USGS exports store Date and Time in separate columns.
if "Date" in df.columns and "Time" in df.columns:
    df["Datetime"] = pd.to_datetime(
        df["Date"].astype(str) + " " + df["Time"].astype(str),
        infer_datetime_format=True, errors="coerce"
    )
elif "Date" in df.columns:
    df["Datetime"] = pd.to_datetime(df["Date"], infer_datetime_format=True,
                                    errors="coerce")
else:
    raise ValueError("CSV has no 'Date' column — check the file structure.")

# Drop rows with unparseable timestamps or missing magnitude
df = df.dropna(subset=["Datetime", "Magnitude"])
df["Magnitude"] = pd.to_numeric(df["Magnitude"], errors="coerce")
df = df.dropna(subset=["Magnitude"])

# Temporal sort — fundamental for any time-series model
df = df.sort_values("Datetime").reset_index(drop=True)

mag = df["Magnitude"]
print(f"  Records : {len(df):,}")
print(f"  Span    : {df['Datetime'].min()} → {df['Datetime'].max()}")
print(f"  Magnitude: min={mag.min():.2f}, max={mag.max():.2f}, "
      f"mean={mag.mean():.2f}, std={mag.std():.2f}")


# ============================================================
# SECTION 3 — TEMPORAL SPLIT (no shuffling, ever)
# ============================================================
print("\n[2/10] Splitting …")
N        = len(df)
n_train  = int(N * TRAIN_RATIO)
n_val    = int(N * VAL_RATIO)
n_test   = N - n_train - n_val

# Strict non-overlapping temporal windows
IDX_TRAIN_END = n_train
IDX_VAL_END   = n_train + n_val

print(f"  Train : rows [0, {IDX_TRAIN_END})    → {n_train:,} samples")
print(f"  Val   : rows [{IDX_TRAIN_END}, {IDX_VAL_END})  → {n_val:,} samples")
print(f"  Test  : rows [{IDX_VAL_END}, {N})   → {n_test:,} samples")


# ============================================================
# SECTION 4 — CAUSAL FEATURE ENGINEERING
# ============================================================
# Philosophy
# ----------
# All operators below are strictly LEFT-SIDED (causal):
#   • ewm(adjust=False)      → α·x_t + (1-α)·EMA_{t-1}
#   • rolling(w).std()       → std of [t-w, t] (backward window)
#   • diff()                 → x_t − x_{t-1}
#   • sosfilt (forward-only) → causal IIR pass
# Computing on the FULL chronological df before splitting is equivalent
# to compute-per-split with cross-boundary warm-up concat, and avoids
# NaN artifacts at split boundaries. The scaler is still fit on training
# rows ONLY — no statistical leakage.

print("\n[3/10] Engineering features …")

m = df["Magnitude"].values
dt = df["Datetime"]

# ── Low-pass filters ──────────────────────────────────────────────────────
# EMA (fast): α ≈ 2/(span+1). Smooth but has exponential impulse response.
ema_fast = pd.Series(m).ewm(span=10, adjust=False).mean().values

# EMA (slow): long-term seismic background level
ema_slow = pd.Series(m).ewm(span=50, adjust=False).mean().values

# Causal Butterworth low-pass (3rd order, Wn=0.05)
# sosfilt processes left→right (causal). Wn=0.05 means 5% of Nyquist,
# i.e., it retains slow cycles and attenuates rapid fluctuations.
# This gives a MUCH steeper roll-off (-60 dB/decade) vs EMA's -20 dB.
sos_lp = butter(3, Wn=0.05, btype="low", analog=False, output="sos")
butter_lp = sosfilt(sos_lp, m)  # causal forward sweep only

# ── High-pass filters ─────────────────────────────────────────────────────
# Subtracting the low-pass isolates seismic transients / short-period shocks.
# Using the Butterworth residual gives a cleaner spectral separation than
# the EMA residual (no exponential tail leakage into the passband).
butter_hp = m - butter_lp      # Butterworth high-pass residual
ema_hp    = m - ema_fast       # EMA high-pass (retained as auxiliary)

# ── Volatility ────────────────────────────────────────────────────────────
# Rolling standard deviation captures local seismic activity clusters.
# min_periods=2 avoids NaN on the very first row; fill 0 (no data yet).
roll_std = (pd.Series(m)
            .rolling(10, min_periods=2)
            .std()
            .fillna(0)
            .values)

# ── Benioff Strain (seismic energy accumulation) ──────────────────────────
# Seismic moment M₀ ∝ 10^(1.5·M). √M₀ (Benioff strain) reveals the
# loading-unloading cycle: it rises steadily then drops after a large rupture.
benioff_inst = np.sqrt(np.power(10.0, 1.5 * m))            # instantaneous
benioff_roll = (pd.Series(benioff_inst)
                .rolling(20, min_periods=1)
                .sum()
                .values)                                      # rolling 20-step sum
# Normalize by 100-step rolling mean → roughly stationary ratio
benioff_base = (pd.Series(benioff_roll)
                .rolling(100, min_periods=1)
                .mean()
                .values + 1e-8)
benioff_norm = benioff_roll / benioff_base

# ── Inter-event Time ──────────────────────────────────────────────────────
# Raw gap in hours; first event has no predecessor → fill 0.
inter_hrs = dt.diff().dt.total_seconds().div(3600).fillna(0).values

# FIX: log1p compresses the heavy Omori-law tail (power-law distribution).
# The original log1p(hours) is correct; we also add a second quantile-based
# feature to capture whether this gap is unusually long (aftershock silence).
log_inter = np.log1p(inter_hrs)

# Rolling quantile: is the current gap longer than the local 75th percentile?
# (Unusually long silences often precede major quakes in some sequences.)
q75 = (pd.Series(inter_hrs)
       .rolling(50, min_periods=5)
       .quantile(0.75)
       .fillna(inter_hrs.mean())
       .values)
inter_ratio = np.log1p(inter_hrs) / (np.log1p(q75) + 1e-8)

# ── Cyclic Temporal Encodings ────────────────────────────────────────────
# Sin/cos encoding preserves temporal topology (23:59 is close to 00:00).
hour = dt.dt.hour.values
dow  = dt.dt.dayofweek.values

sin_hour = np.sin(2 * np.pi * hour / 24)
cos_hour = np.cos(2 * np.pi * hour / 24)
sin_dow  = np.sin(2 * np.pi * dow  / 7)
cos_dow  = np.cos(2 * np.pi * dow  / 7)

# ── Assemble feature matrix ───────────────────────────────────────────────
FEATURE_NAMES = [
    "Magnitude",      # target column kept as input feature (autoregressive)
    "ema_fast",       # EMA-10 low-pass
    "ema_slow",       # EMA-50 low-pass
    "butter_lp",      # Butterworth low-pass
    "butter_hp",      # Butterworth high-pass (transients)
    "ema_hp",         # EMA high-pass (auxiliary)
    "roll_std",       # local volatility
    "benioff_norm",   # normalized Benioff strain
    "log_inter",      # log inter-event hours
    "inter_ratio",    # gap relative to local baseline
    "sin_hour",       # cyclic hour-of-day
    "cos_hour",
    "sin_dow",        # cyclic day-of-week
    "cos_dow",
]
N_FEAT = len(FEATURE_NAMES)

feature_matrix = np.column_stack([
    m, ema_fast, ema_slow, butter_lp, butter_hp, ema_hp,
    roll_std, benioff_norm, log_inter, inter_ratio,
    sin_hour, cos_hour, sin_dow, cos_dow,
]).astype(np.float32)

# Sanity check: no NaNs / Infs should escape
assert np.all(np.isfinite(feature_matrix)), (
    "Feature matrix contains NaN/Inf — check input data for gaps or "
    "corrupted magnitude values."
)
print(f"  Feature matrix: {feature_matrix.shape}  — {N_FEAT} features, no NaN/Inf")


# ============================================================
# SECTION 5 — SCALING (fit on training rows ONLY)
# ============================================================
print("\n[4/10] Scaling …")

# RobustScaler uses median + IQR, so extreme magnitudes do not compress
# the rest of the feature space (unlike MinMaxScaler or StandardScaler).
scaler = RobustScaler()
scaler.fit(feature_matrix[:IDX_TRAIN_END])     # ← training rows ONLY

train_scaled = scaler.transform(feature_matrix[:IDX_TRAIN_END]).astype(np.float32)
val_scaled   = scaler.transform(feature_matrix[IDX_TRAIN_END:IDX_VAL_END]).astype(np.float32)
test_scaled  = scaler.transform(feature_matrix[IDX_VAL_END:]).astype(np.float32)


# ============================================================
# SECTION 6 — SEQUENCE CREATION WITH CROSS-SPLIT WARM-UP
# ============================================================
# FIX: Original per-split create_sequences() discarded the first SEQ_LEN
# samples of val and test because it couldn't form a full look-back window
# within the split alone. At deployment, the model WILL use training
# observations as context for the very first predictions — so the
# evaluation should mirror that.
#
# Solution: prepend the last SEQ_LEN rows of the PRECEDING split as
# warm-up context. These rows are already correctly scaler-transformed
# using training statistics, so there is zero statistical leakage.
# The TARGET always comes from the CURRENT split only.

print("\n[5/10] Creating sequences …")

MAG_IDX = 0   # Magnitude is column 0 of the feature matrix

def make_train_sequences(data: np.ndarray, seq_len: int):
    """Standard overlapping sequences — no warm-up needed for training.

    sliding_window_view was removed: it returns (n-seq_len+1, 1, seq_len, n_feat)
    and the size-1 axis is axis=1, NOT axis=2 as originally written, causing
    ValueError at squeeze time. np.stack is explicit, unambiguous, and equally
    fast for datasets of this size.
    """
    n_seq = len(data) - seq_len
    X = np.stack([data[i : i + seq_len] for i in range(n_seq)], axis=0)
    y = data[seq_len:, MAG_IDX]                 # targets: magnitude column
    return X.astype(np.float32), y.astype(np.float32)

def make_eval_sequences(prior_tail: np.ndarray,
                         current: np.ndarray,
                         seq_len: int):
    """
    Generates one sequence per row of `current`.
    The first seq_len sequences draw context from `prior_tail` (last SEQ_LEN
    rows of the preceding split), seamlessly continuing the look-back window.

    Args
    ----
    prior_tail : (seq_len, n_feat) — scaled tail of previous split
    current    : (n, n_feat)       — scaled data of the current split
    seq_len    : int

    Returns
    -------
    X : (n, seq_len, n_feat)
    y : (n,)
    """
    assert prior_tail.shape[0] == seq_len, (
        f"prior_tail must have exactly {seq_len} rows, got {prior_tail.shape[0]}"
    )
    padded = np.concatenate([prior_tail, current], axis=0)   # (seq_len+n, n_feat)
    n = len(current)
    X = np.stack([padded[i:i+seq_len] for i in range(n)], axis=0)
    y = current[:, MAG_IDX]
    return X.astype(np.float32), y.astype(np.float32)

X_train, y_train = make_train_sequences(train_scaled, SEQ_LEN)
X_val,   y_val   = make_eval_sequences(train_scaled[-SEQ_LEN:], val_scaled,  SEQ_LEN)
X_test,  y_test  = make_eval_sequences(val_scaled[-SEQ_LEN:],   test_scaled, SEQ_LEN)

print(f"  Train : {X_train.shape}  |  Val: {X_val.shape}  |  Test: {X_test.shape}")
print(f"  (Val/Test now cover 100% of their splits — no boundary loss)")


# ============================================================
# SECTION 7 — SAMPLE WEIGHTING (numerically stable)
# ============================================================
# FIX: Original used exp(M - M_min). For a dataset with M_min=4, M_max=9:
#   exp(9-4) = exp(5) ≈ 148 →  M9 quake gets 148× the gradient signal.
# This destabilises Adam (extremely high variance in the gradient estimate).
#
# Fix: log-linear weighting. On the Richter scale (which is itself already
# log-energy), a linear weight is still super-linear in physical energy,
# but the RATIO between largest and smallest weight stays manageable.
#
#   w_i = log1p(M_i - M_min + 1)  →  weight range ≈ [1.1, 2.4] for M∈[4,9]
#
# Weights are normalised to mean=1 so the effective learning rate is
# unchanged on average (only the relative emphasis shifts).

print("\n[6/10] Computing sample weights …")

def inv_transform_magnitude(scaled_col: np.ndarray) -> np.ndarray:
    """Inverse-transform magnitude-only column using the fitted scaler."""
    dummy = np.zeros((len(scaled_col), N_FEAT), dtype=np.float32)
    dummy[:, MAG_IDX] = scaled_col
    return scaler.inverse_transform(dummy)[:, MAG_IDX]

y_train_mag = inv_transform_magnitude(y_train)
M_min_train = y_train_mag.min()

# Soft log-linear weight
raw_w = np.log1p(y_train_mag - M_min_train + 1.0)
sample_weights = (raw_w / raw_w.mean()).astype(np.float32)

print(f"  Magnitude range in train : [{M_min_train:.2f}, {y_train_mag.max():.2f}]")
print(f"  Weight range             : [{sample_weights.min():.3f}, {sample_weights.max():.3f}]")


# ============================================================
# SECTION 8 — MODEL (Attention-augmented BiLSTM)
# ============================================================
print("\n[7/10] Building model …")

def build_model(seq_len: int, n_features: int,
                l2_reg: float = L2_REG,
                dropout: float = DROPOUT) -> Model:
    """
    Architecture:
      Conv1D (causal padding) → BatchNorm → Conv1D → BN → MaxPool → Dropout
      → BiLSTM (return_sequences=True)
      → Multi-Head Self-Attention + residual + LayerNorm
      → BiLSTM (return_sequences=False)
      → Dense head → scalar prediction

    Key design choices:
    • padding="causal" in Conv1D: no future timesteps leak into receptive field.
    • MultiHeadAttention over the BiLSTM output: captures non-local temporal
      dependencies (e.g., aftershock sequences days apart).
    • Residual + LayerNorm: stabilises deep networks and improves gradient flow.
    • Huber loss: quadratic for small errors, linear for large — provides stable
      gradient on both minor tremors AND extreme events.
    """
    reg = l2(l2_reg)

    inp = Input(shape=(seq_len, n_features), name="input")

    # ── Temporal CNN: extract local seismic patterns ───────────────────────
    # FIX: padding="causal" (original used "same" which is non-causal)
    x = Conv1D(64, 3, padding="causal", activation="relu",
               kernel_regularizer=reg, name="conv1")(inp)
    x = BatchNormalization()(x)
    x = Conv1D(64, 3, padding="causal", activation="relu",
               kernel_regularizer=reg, name="conv2")(x)
    x = BatchNormalization()(x)
    x = MaxPooling1D(2)(x)
    x = Dropout(dropout)(x)

    # ── BiLSTM 1: bidirectional temporal dynamics ──────────────────────────
    x = Bidirectional(
        LSTM(64, return_sequences=True, kernel_regularizer=reg),
        name="bilstm1"
    )(x)
    x = Dropout(dropout)(x)

    # ── Multi-Head Self-Attention with residual connection ─────────────────
    # 4 heads × key_dim=16 → total query space = 64 (matches BiLSTM output)
    # Attends over the sequence to capture long-range magnitude dependencies
    # (e.g. a large foreshock days earlier influencing current predictions).
    attn_out = MultiHeadAttention(
        num_heads=4, key_dim=16, dropout=dropout/2, name="mha"
    )(x, x, x)
    x = LayerNormalization(epsilon=1e-6)(Add(name="residual_attn")([x, attn_out]))

    # ── BiLSTM 2: compress attended sequence to a single vector ───────────
    x = Bidirectional(
        LSTM(32, return_sequences=False, kernel_regularizer=reg),
        name="bilstm2"
    )(x)
    x = Dropout(dropout)(x)

    # ── Dense projection ───────────────────────────────────────────────────
    x = Dense(32, activation="relu", kernel_regularizer=reg)(x)
    x = Dense(16, activation="relu")(x)
    out = Dense(1, name="magnitude")(x)

    model = Model(inp, out, name="EQ_BiLSTM_Attention")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=LR),
        loss=Huber(delta=HUBER_DELTA),
        metrics=["mae"]
    )
    return model

model = build_model(SEQ_LEN, N_FEAT)
model.summary(line_length=90)


# ============================================================
# SECTION 9 — TRAINING
# ============================================================
print("\n[8/10] Training …")

callbacks = [
    EarlyStopping(
        monitor="val_loss", patience=25,
        restore_best_weights=True, verbose=1
    ),
    ReduceLROnPlateau(
        monitor="val_loss", factor=0.5,
        patience=10, min_lr=1e-6, verbose=1
    ),
]

history = model.fit(
    X_train, y_train,
    sample_weight=sample_weights,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_data=(X_val, y_val),
    callbacks=callbacks,
    # FIX: shuffle=False preserves temporal mini-batch ordering.
    # The look-back window inside X already encodes temporal context;
    # shuffling (X,y) pairs does not break sequences themselves, but
    # temporal mini-batch order is more consistent with time-series logic.
    shuffle=False,
    verbose=1,
)

stopped_epoch = len(history.history["loss"])
print(f"\n  Stopped at epoch {stopped_epoch} "
      f"(best val_loss = {min(history.history['val_loss']):.5f})")


# ============================================================
# SECTION 10 — EVALUATION
# ============================================================
print("\n[9/10] Evaluating …")

pred_scaled = model.predict(X_test, verbose=0).ravel()
pred_mag    = inv_transform_magnitude(pred_scaled)
actual_mag  = inv_transform_magnitude(y_test)

def evaluate_split(actual: np.ndarray, pred: np.ndarray, label: str) -> dict:
    """
    Metrics:
      MAE, RMSE, R²    — standard regression metrics
      Bias             — systematic over/under-prediction
      DirAcc           — directional accuracy (sign of Δmagnitude)
                         High directional accuracy = useful for hazard trends.
    """
    if len(actual) < 5:
        print(f"  [{label}] Skipped — only {len(actual)} samples")
        return {}
    mae     = mean_absolute_error(actual, pred)
    rmse    = np.sqrt(mean_squared_error(actual, pred))
    r2      = r2_score(actual, pred)
    bias    = (pred - actual).mean()
    dir_acc = (np.sign(np.diff(actual)) == np.sign(np.diff(pred))).mean()

    print(f"\n  ── {label} (n={len(actual)}) ──")
    print(f"     MAE  = {mae:.4f}   RMSE = {rmse:.4f}   R² = {r2:.4f}")
    print(f"     Bias = {bias:+.4f}   DirAcc = {dir_acc:.3%}")
    return dict(mae=mae, rmse=rmse, r2=r2, bias=bias, dir_acc=dir_acc)

print("\n  ══ TEST SET ══")
metrics_all = evaluate_split(actual_mag, pred_mag, "All magnitudes")

# Stratified evaluation — the critical question: how does the model perform
# on DANGEROUS earthquakes vs. the common small background seismicity?
bins = [
    (0.0, 4.0, "< 4.0  (minor)"),
    (4.0, 5.0, "4.0–5.0 (light)"),
    (5.0, 6.0, "5.0–6.0 (moderate)"),
    (6.0, 7.0, "6.0–7.0 (strong)"),
    (7.0, float("inf"), "≥ 7.0  (major)"),
]
bin_metrics = {}
for lo, hi, lbl in bins:
    mask = (actual_mag >= lo) & (actual_mag < hi)
    bin_metrics[lbl] = evaluate_split(actual_mag[mask], pred_mag[mask], lbl)


# ============================================================
# SECTION 11 — DIAGNOSTIC VISUALISATION
# ============================================================
print("\n[10/10] Plotting …")

fig, axes = plt.subplots(2, 2, figsize=(15, 11))
fig.suptitle("Earthquake Magnitude Forecast — Diagnostics", fontsize=14, y=1.01)

# ── (a) Training curves ────────────────────────────────────────────────────
ax = axes[0, 0]
epochs_x = range(1, stopped_epoch + 1)
ax.plot(epochs_x, history.history["loss"],     label="Train Huber loss", lw=1.5)
ax.plot(epochs_x, history.history["val_loss"], label="Val Huber loss",   lw=1.5)
ax.set_title("(a) Training Curves")
ax.set_xlabel("Epoch")
ax.set_ylabel("Huber Loss")
ax.legend()
ax.grid(alpha=0.3)

# ── (b) Actual vs Predicted (scatter) ─────────────────────────────────────
ax = axes[0, 1]
sc = ax.scatter(actual_mag, pred_mag, c=actual_mag, cmap="plasma",
                alpha=0.35, s=8, rasterized=True)
lims = [min(actual_mag.min(), pred_mag.min()) - 0.2,
        max(actual_mag.max(), pred_mag.max()) + 0.2]
ax.plot(lims, lims, "r--", lw=1.2, label="Perfect prediction")
ax.set_xlabel("Actual Magnitude")
ax.set_ylabel("Predicted Magnitude")
ax.set_title("(b) Actual vs Predicted (colour = actual M)")
ax.legend(fontsize=9)
ax.set_xlim(lims); ax.set_ylim(lims)
plt.colorbar(sc, ax=ax, label="Magnitude")
ax.grid(alpha=0.3)

# ── (c) Residuals vs Actual (heteroscedasticity check) ────────────────────
ax = axes[1, 0]
residuals = pred_mag - actual_mag
ax.scatter(actual_mag, residuals, c=actual_mag, cmap="plasma",
           alpha=0.35, s=8, rasterized=True)
ax.axhline(0, color="red", lw=1.2, linestyle="--")
ax.set_xlabel("Actual Magnitude")
ax.set_ylabel("Residual  (Predicted − Actual)")
ax.set_title("(c) Residuals vs Actual")
# Annotate bias per bin
for lo, hi, lbl in bins:
    mask = (actual_mag >= lo) & (actual_mag < hi)
    if mask.sum() >= 5:
        mid = (lo + min(hi, actual_mag.max())) / 2
        b = (pred_mag[mask] - actual_mag[mask]).mean()
        ax.annotate(f"bias={b:+.2f}", xy=(mid, b),
                    fontsize=7, ha="center",
                    bbox=dict(boxstyle="round,pad=0.2", fc="white", alpha=0.7))
ax.grid(alpha=0.3)

# ── (d) Time-series window (last 300 samples) ─────────────────────────────
ax = axes[1, 1]
n_plot = min(300, len(actual_mag))
ax.plot(actual_mag[-n_plot:], label="Actual",    alpha=0.75, lw=1.0)
ax.plot(pred_mag[-n_plot:],   label="Predicted", alpha=0.75, lw=1.0, ls="--")
ax.set_title(f"(d) Last {n_plot} Test-Set Predictions")
ax.set_xlabel("Sample index (within test set)")
ax.set_ylabel("Magnitude")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
out_fig = "/mnt/user-data/outputs/earthquake_diagnostics.png"
os.makedirs(os.path.dirname(out_fig), exist_ok=True)
plt.savefig(out_fig, dpi=150, bbox_inches="tight")
print(f"  Saved → {out_fig}")
plt.close()


# ============================================================
# SUMMARY OF ALL FIXES (printed at end for quick reference)
# ============================================================
print("""
╔══════════════════════════════════════════════════════════════════╗
║              FIX SUMMARY vs ORIGINAL CODE                       ║
╠══════════════════════════════════════════════════════════════════╣
║ [1] ENTROPY POOL  : TF_DETERMINISTIC_OPS + TF_CUDNN_DETERMINISTIC ║
║                    set BEFORE TF import; try/except guard added  ║
║ [2] NO LEAKAGE    : Scaler fit on train rows only. All feature   ║
║                    operators are provably causal (left-sided).   ║
║ [3] LOW-PASS      : 3rd-order causal Butterworth IIR added       ║
║                    (−60 dB/decade roll-off vs EMA's −20 dB)     ║
║ [4] HIGH-PASS     : Butterworth residual (cleaner band-split)    ║
║ [5] INTER-EVENT   : log1p(hrs) + inter_ratio + sin/cos cyclic   ║
║                    hour-of-day & day-of-week encodings           ║
║ [6] BENIOFF STRAIN: √(10^1.5M) rolling sum → energy loading     ║
║ [7] SEQUENCES     : Cross-split warm-up — 100% val/test coverage  ║
║                    (original lost SEQ_LEN samples per boundary)  ║
║ [8] WEIGHTS       : log-linear (max ~2.4×) vs exp (max ~148×)   ║
║ [9] CONV PADDING  : causal (original used non-causal "same")     ║
║ [10] ATTENTION    : Multi-Head Self-Attention + residual+LN      ║
║ [11] SHUFFLE=FALSE: temporal mini-batch ordering preserved       ║
║ [12] EVALUATION   : Bias, DirAcc, per-bin stratified metrics,    ║
║                    4-panel diagnostic figure                     ║
╚══════════════════════════════════════════════════════════════════╝
""")
```
