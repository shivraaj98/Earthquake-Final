```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Input, Conv1D, MaxPooling1D, Bidirectional, LSTM,
                                     Dense, Dropout, Flatten, Attention as KerasAttention,
                                     Add, Permute, Reshape, Multiply, Lambda)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras import backend as K
from pathlib import Path
import os
import sys

def in_colab() -> bool:
    return "google.colab" in sys.modules or "COLAB_RELEASE_TAG" in os.environ

SEED = 42

def load_and_preprocess(csv_filename):
    possible_paths = [
        Path(csv_filename),
        Path("/content") / csv_filename,
        Path("/content/earthquake (1).csv"),
    ]
    resolved_path = None
    for path in possible_paths:
        if path.exists():
            resolved_path = path.resolve()
            break
    if resolved_path is None:
        if in_colab():
            from google.colab import files
            uploaded = files.upload()
            first_name = next(iter(uploaded.keys()))
            resolved_path = Path(first_name).resolve()
        else:
            raise FileNotFoundError(f"File {csv_filename} not found.")
    print(f"Using data file: {resolved_path}")
    df = pd.read_csv(resolved_path)
    df['Date'] = pd.to_datetime(df['Date'])
    df.sort_values('Date', inplace=True)
    df.reset_index(drop=True, inplace=True)
    feature_cols = ['Magnitude', 'Depth', 'Longitude', 'Latitude']
    target_col = 'Magnitude'
    df = df.dropna(subset=feature_cols + [target_col])
    X_raw = df[feature_cols].values
    y_raw = df[target_col].values
    dates = df['Date'].values
    return X_raw, y_raw, dates, feature_cols

def create_sequences(X, y, seq_len):
    X_seq, y_seq = [], []
    for i in range(len(X) - seq_len):
        X_seq.append(X[i:i+seq_len])
        y_seq.append(y[i+seq_len])
    return np.array(X_seq), np.array(y_seq)

class AdditiveAttention(tf.keras.layers.Layer):
    def __init__(self, units, **kwargs):
        super(AdditiveAttention, self).__init__(**kwargs)
        self.units = units
        self.W1 = Dense(units, activation='tanh')
        self.W2 = Dense(1)
    def call(self, inputs):
        score = self.W1(inputs)
        attention_weights = tf.nn.softmax(self.W2(score), axis=1)
        context = attention_weights * inputs
        context = tf.reduce_sum(context, axis=1)
        return context
    def get_config(self):
        config = super().get_config()
        config.update({'units': self.units})
        return config

def build_model(seq_len, n_features, cnn_filters=32, lstm_units=64, dropout_rate=0.3):
    inputs = Input(shape=(seq_len, n_features))
    x = Conv1D(filters=cnn_filters, kernel_size=2, activation='relu', padding='same')(inputs)
    x = MaxPooling1D(pool_size=2)(x)
    x = Conv1D(filters=cnn_filters*2, kernel_size=2, activation='relu', padding='same')(x)
    x = MaxPooling1D(pool_size=2)(x)
    x = Bidirectional(LSTM(lstm_units, return_sequences=True, dropout=dropout_rate))(x)
    x = Bidirectional(LSTM(lstm_units//2, return_sequences=True, dropout=dropout_rate))(x)
    attention_out = AdditiveAttention(units=lstm_units)(x)
    x = Dense(32, activation='relu')(attention_out)
    x = Dropout(dropout_rate)(x)
    outputs = Dense(1, activation='linear')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(learning_rate=0.001), loss='mse', metrics=['mae'])
    return model

# Run logic in global scope
csv_file = "earthquakeisc.csv"
X_raw, y_raw, dates, feature_cols = load_and_preprocess(csv_file)
split_date = np.datetime64('2021-01-01')
train_mask = dates < split_date
test_mask = (dates >= split_date) & (dates < np.datetime64('2024-01-01'))
X_train_raw = X_raw[train_mask]
y_train_raw = y_raw[train_mask]
X_test_raw = X_raw[test_mask]
y_test_raw = y_raw[test_mask]
dates_test = dates[test_mask]
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)
SEQ_LEN = 10
X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_raw, SEQ_LEN)
X_test_seq, y_test_seq = create_sequences(X_test_scaled, y_test_raw, SEQ_LEN)
dates_test_seq = dates_test[SEQ_LEN:]
model = build_model(seq_len=SEQ_LEN, n_features=X_train_seq.shape[2])
early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)
model.fit(X_train_seq, y_train_seq, validation_split=0.2, epochs=100, batch_size=32, callbacks=[early_stop, reduce_lr], verbose=1)
y_pred_seq = model.predict(X_test_seq).flatten()
plt.figure(figsize=(10,6))
plt.plot(dates_test_seq, y_test_seq, label='Actual')
plt.plot(dates_test_seq, y_pred_seq, label='Predicted')
plt.legend()
plt.show()
```
